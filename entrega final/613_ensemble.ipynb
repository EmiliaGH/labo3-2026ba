Promedio simple de los 2 mejores submits ya generados: LGB07_WF (0.251) y LGB06_FE609 (0.256). No reentrena nada -- usa las predicciones ya calculadas.

```python
import os
import pandas as pd
```

```python
PARAM = {
    'experimento': 'ENS01',
    'kaggle_competition': 'labo-iii-2026-ba',
    'submits_a_promediar': [
        '/home/ds/exp/LGB07_WF/LGB07_WF_submit.csv',
        '/home/ds/exp/LGB06_FE609/LGB06_FE609_submit.csv',
    ]
}

ruta = os.path.join('/home/ds/exp', PARAM['experimento'])
os.makedirs(ruta, exist_ok=True)
print(ruta)
```

## 1. Verificar que los archivos existan antes de seguir

```python
for p in PARAM['submits_a_promediar']:
    print(p, os.path.isfile(p))
```

## 2. Cargar y promediar
Cada submit ya viene filtrado a los productos a predecir, mismo orden de columnas (`product_id, tn`).

```python
tablas = [pd.read_csv(p) for p in PARAM['submits_a_promediar']]

base = tablas[0][["product_id"]].copy()
for i, t in enumerate(tablas):
    base[f"tn_{i}"] = t["tn"].values

cols_tn = [c for c in base.columns if c.startswith("tn_")]
base["tn"] = base[cols_tn].mean(axis=1)

submit = base[["product_id", "tn"]]
print(submit.shape)
submit.head()
```

## 3. Guardar y submit

```python
archivo_submit = os.path.join(ruta, f"{PARAM['experimento']}_submit.csv")
submit.to_csv(archivo_submit, index=False)
print(archivo_submit)
```

```python
def kaggle_submit(competencia, archivo, mensaje):
    comando = f'kaggle competitions submit -c {competencia} -f {archivo} -m "{mensaje}"'
    os.system(comando)

kaggle_submit(PARAM['kaggle_competition'], archivo_submit, f"{PARAM['experimento']} ensamble LGB07_WF + LGB06_FE609")
```
